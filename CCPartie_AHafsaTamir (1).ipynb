### Présentation
*Q1 **Noms** : Régression Ridge (Ridge Regression / Tikhonov Regularization).
* **Famille** : Régression (Apprentissage supervisé).
* **Objectif** : Ajuster un modèle de régression linéaire tout en limitant la magnitude des coefficients pour éviter le surapprentissage.
* **Intuition & Étapes** : **Entrée** : Variables $X$ et cible $y$. $\rightarrow$ **Apprentissage** : Ajustement des poids en introduisant une pénalité quadratique sur la norme $L_2$ des coefficients. $\rightarrow$ **Sortie** : Prédictions continues $\hat{y}$ moins sensibles aux variations des variables d'entrée.
* **Critère mathématique** :
  $$J(w, b) = \frac{1}{N} \sum_{i=1}^{N} (y_i - (w^T x_i + b))^2 + \alpha \|w\|_2^2$$
  * *Paramètres* : Coefficients (poids $w$) et intercept $b$.
  * *Hyperparamètres* : Coefficient de régularisation $\alpha$ (ou $\lambda$). Plus $\alpha$ est élevé, plus les poids sont contraints vers zéro.
* **Avantages** : Réduit efficacement la variance du modèle en présence de multicolinéarité ; assure une solution stable et unique.
* **Limites** : Ne fait pas de sélection de variables (les coefficients tendent vers 0 mais ne s'annulent jamais) ; nécessite la mise à l'échelle préalable des variables (StandardScaler).

* **Cas d'usage 2 (Marketing)** : Modélisation de l'élasticité-prix de ventes en présence de nombreux canaux de campagnes publicitaires simultanés.

## 3. Lasso

### Présentation
* **Noms** : Régression Lasso (Lasso Regression / Least Absolute Shrinkage and Selection Operator).
* **Famille** : Régression (Apprentissage supervisé).
* **Objectif** : Ajuster un modèle linéaire tout en appliquant une pénalité qui favorise la parcimonie (annulation exacte de certains coefficients).
* **Intuition & Étapes** : **Entrée** : Variables $X$ et cible $y$. $\rightarrow$ **Apprentissage** : Minimisation des résidus avec une pénalité sur la somme des valeurs absolues des coefficients (norme $L_1$). $\rightarrow$ **Sortie** : Prédictions continues et sélection naturelle de variables en mettant les coefficients des variables redondantes à exactement zéro.
* **Critère mathématique** :
  $$J(w, b) = \frac{1}{2N} \sum_{i=1}^{N} (y_i - (w^T x_i + b))^2 + \alpha \|w\|_1$$
  * *Paramètres* : Coefficients (poids $w$) et intercept $b$.
  * *Hyperparamètres* : Paramètre de régularisation $\alpha \ge 0$. Plus $\alpha$ est grand, plus le nombre de coefficients mis à zéro est élevé.
* **Avantages** : Effectue une sélection automatique de caractéristiques (feature selection) ; génère des modèles très interprétables et légers.
* **Limites** : Si plusieurs caractéristiques sont fortement corrélées, Lasso a tendance à n'en choisir qu'une seule de manière arbitraire ; perd en performance si le nombre de variables utiles est supérieur au nombre d'observations ($p > n$).

### Schéma explicatif
Le schéma ci-dessous affiche l'évolution de la valeur absolue des coefficients appris pour montrer comment la régularisation $L_1$ parvient à forcer certains poids à être rigoureusement égaux à zéro.

### Contextes d'utilisation
* **Cas d'usage 1 (Finance/Économie)** : Sélection d'un sous-ensemble optimal d'indicateurs macroéconomiques pertinents pour prédire la croissance du PIB parmi des centaines de candidats.
* **Cas d'usage 2 (Marketing)** : Identification des canaux d'acquisition de trafic web les plus rentables et élimination des métriques de bruit non significatives.
* **Inapproprié quand** : Lorsque toutes les variables d'entrée ont un effet réel de faible intensité et que l'on ne souhaite en éliminer aucune (Ridge est alors préférable).
* **Métriques adaptées** : Coefficient de détermination $R^2$, Erreur quadratique moyenne (MSE), Erreur absolue moyenne (MAE).

### Analyse de l'expérience Python
* *Entrées* : Variables explicatives standardisées à l'aide de `StandardScaler`.
* *Hyperparamètres* : Coefficient de pénalité $\alpha = 0.1$ et un nombre maximal d'itérations fixé à $10\ 000$.
* *Sortie obtenue* : Le coefficient de détermination $R^2$ sur le jeu de test. Un score proche de 1 valide la qualité prédictive du modèle parcimonieux.

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import Lasso, LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score

# Entraînement Lasso
modele_lasso = make_pipeline(StandardScaler(), Lasso(alpha=0.1, max_iter=10000))
modele_lasso.fit(Xr_ent, yr_ent)
pred_lasso = modele_lasso.predict(Xr_test)
r2_lasso = r2_score(yr_test, pred_lasso)
print('R² Lasso sur le jeu de test:', r2_lasso)

# Coefficients
coef_lasso = modele_lasso.named_steps['lasso'].coef_
print('Coefficients Lasso :', coef_lasso)

# Tracé du schéma explicatif
plt.figure(figsize=(8, 4))
x_indices = np.arange(len(coef_lasso))
plt.bar(x_indices, coef_lasso, color='purple', alpha=0.7, label='Coefficients Lasso (L1)')
plt.axhline(0, color='black', linewidth=1, linestyle='--')
plt.xlabel('Index des caractéristiques (Features)')
plt.ylabel('Valeur du coefficient')
plt.title('Schéma explicatif : Sélection de variables par Lasso (certains coefficients s\'annulent)')
plt.xticks(x_indices)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 4. Régression polynomiale

### Présentation
* **Noms** : Régression polynomiale (Polynomial Regression).
* **Famille** : Régression (Apprentissage supervisé).
* **Objectif** : Modéliser des relations non linéaires complexes entre les variables explicatives et la variable cible en introduisant des termes de degrés supérieurs (interactions et puissances).
* **Intuition & Étapes** : **Entrée** : Variables de dimension $d$. $
ightarrow$ **Calcul** : Transformation des caractéristiques initiales en combinant leurs puissances (ex: $x_1, x_1^2, x_1 x_2$). $
ightarrow$ **Apprentissage** : Application d'un estimateur linéaire sur ces nouvelles variables étendues. $
ightarrow$ **Sortie** : Une courbe ou surface d'ajustement non linéaire dans l'espace original.
* **Critère mathématique** :
  $$y_i = w_0 + w_1 x_i + w_2 x_i^2 + ... + w_d x_i^d + \epsilon$$
  La minimisation se fait par les moindres carrés (MSE) sur l'espace transformé.
  * *Paramètres* : Coefficients associés aux puissances $w_j$ et le biais.
  * *Hyperparamètres* : Le degré du polynôme (`degree`) et les paramètres du régularisateur associé (ex: $\alpha$ de Ridge).
* **Avantages** : Permet de capter des motifs courbés et non linéaires sans quitter le cadre mathématique robuste de la régression linéaire ; flexible.
* **Limites** : Sensible au surapprentissage (overfitting) dès que le degré augmente ; instable aux frontières du domaine des données (phénomène de Runge).

### Schéma explicatif
Le graphique ci-dessous illustre l'ajustement non linéaire réalisé grâce à l'expansion polynomiale de degré 2 sur une dimension de nos données.

### Contextes d'utilisation
* **Cas d'usage 1 (Finance/Énergie)** : Modélisation de la relation non linéaire entre la température extérieure et la consommation d'électricité (courbe en U).
* **Cas d'usage 2 (Gestion des stocks/Logistique)** : Évaluation du coût total d'approvisionnement en fonction du volume commandé (effets d'échelle non linéaires).
* **Inapproprié quand** : Les données présentent des sauts brusques ou des discontinuités (les arbres de décision ou forêts aléatoires sont alors plus adaptés).
* **Métriques adaptées** : Erreur absolue moyenne (MAE), Erreur quadratique moyenne (MSE), $R^2$.

### Analyse de l'expérience Python
* *Entrées* : Caractéristiques transformées en termes quadratiques (degré 2) sans biais initial, puis standardisées.
* *Hyperparamètres* : Degré = 2, Régularisation Ridge $\alpha = 10$.
* *Sortie* : Score MAE sur le jeu de test. Permet de juger si la complexité polynomiale réduit l'erreur par rapport au cas purement linéaire.

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_absolute_error

# Entraînement
modele_poly = make_pipeline(
    StandardScaler(),
    PolynomialFeatures(degree=2, include_bias=False),
    Ridge(alpha=10)
)
modele_poly.fit(Xr_ent, yr_ent)
pred_poly = modele_poly.predict(Xr_test)
mae_poly = mean_absolute_error(yr_test, pred_poly)
print('MAE Régression Polynomiale sur le jeu de test:', mae_poly)

# Tracé du schéma explicatif (Projection univariée)
plt.figure(figsize=(8, 5))
x_visu = Xr_ent[:, 0]
y_visu = yr_ent
sort_idx = np.argsort(x_visu)

# Ajustement d'un polynôme de degré 2 sur la première dimension pour visualisation
pipe_visu = make_pipeline(PolynomialFeatures(degree=2), Ridge(alpha=10))
pipe_visu.fit(x_visu.reshape(-1, 1), y_visu)
x_range = np.linspace(x_visu.min(), x_visu.max(), 200).reshape(-1, 1)

plt.scatter(x_visu, y_visu, color='blue', alpha=0.5, label='Données (Dim 0)')
plt.plot(x_range, pipe_visu.predict(x_range), color='darkorange', linewidth=3, label='Ajustement polynomial (Degré 2)')
plt.title('Schéma explicatif : Régression Polynomiale de degré 2')
plt.xlabel('Variable X[0]')
plt.ylabel('Cible y')
plt.legend()
plt.grid(True)
plt.show()
```

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

## Annexes — implémentations manuelles des trois algorithmes imposés

```python
# Régression linéaire par descente de gradient : données univariées
x=np.linspace(-1,1,100); y=2+3*x
poids=0.;biais=0.;pas=.1
for iteration in range(500):
    prediction=poids*x+biais
    erreur=prediction-y
    poids-=pas*2*np.mean(erreur*x)
    biais-=pas*2*np.mean(erreur)
print('Paramètres appris manuellement:',poids,biais)
from sklearn.linear_model import LinearRegression
print('Référence sklearn:',LinearRegression().fit(x.reshape(-1,1),y).coef_,LinearRegression().fit(x.reshape(-1,1),y).intercept_)
```

```python
# K-means manuel : assignation et recalcul jusqu'à stabilité
valeurs=X_clu.copy(); centroids=valeurs[np.random.default_rng(GRAINE).choice(len(valeurs),size=3,replace=False)].copy()
for iteration in range(100):
    distances=((valeurs[:,None,:]-centroids[None,:,:])**2).sum(axis=2)
    groupes=distances.argmin(axis=1)
    nouveaux=np.array([valeurs[groupes==k].mean(axis=0) if (groupes==k).any() else centroids[k] for k in range(3)])
    if np.allclose(nouveaux,centroids):break
    centroids=nouveaux
print('Centroïdes manuels:',centroids)
from sklearn.cluster import KMeans
print('Centroïdes sklearn (ordre arbitraire):',KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(valeurs).cluster_centers_)
```
